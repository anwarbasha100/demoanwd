```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("SalesETL").getOrCreate()

df = spark.read.csv("/Volumes/workspace/default/salesdata", header = True,inferSchema = True)
df.show(1)
```

```python
df.write.mode("overwrite").parquet("/Volumes/workspace/default/salesdata/sales_parquet")
```

```python
parquet_df = spark.read.parquet("/Volumes/workspace/default/salesdata/sales_parquet")
parquet_df.show(3)
```

```python
from pyspark.sql.functions import col
```

```python
parquet_df.select("ORDERNUMBER", "SALES").show(3)

```

```python
parquet_df.filter(parquet_df.STATUS == "NULL").show(3)
```

```python
parquet_df.printSchema()
```

```python
from pyspark.sql.functions import col
```

```python
parquet_df.select("MONTH_ID","YEAR_ID").where(col("COUNTRY") == "USA").show()
```

```python
parquet_df.select("CITY","CUSTOMERNAME").where(col("COUNTRY")== "USA").show()
```
